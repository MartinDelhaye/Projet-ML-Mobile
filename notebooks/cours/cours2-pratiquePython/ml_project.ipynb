```python
#datasets de synthèse
from sklearn.datasets import make_moons , make_circles
# pour le tracé des graphiques
import matplotlib.pyplot as plt
# pour gérer les couleurs
from matplotlib.colors import ListedColormap
# apprentissage
from sklearn import svm
# calcul numérique
import numpy as np
```

```python
X, y = make_moons(noise=0.3, random_state=0)

print(X.shape)
print(y.shape)
print(X)
print(y)
```

```python
# colormap perso
my_cmap = ListedColormap (['#000000', '#FF0000'])
# tracer les points d'apprentissage
plt.scatter (X[:, 0], X[:, 1], c=y, cmap=my_cmap, edgecolors='gray')
plt.xlabel (("x_1"))
plt.ylabel (("x_2"))
plt.legend()

plt.show()
```

```python
# Lancer l’apprentissage avec SVM linéaire
classifier = svm.SVC(kernel='linear', C=10.0)
classifier.fit(X,y)
```

```python
# deux nouveaux échantillons
X_test = np.array([[1.47, -0.99],[-0.55, 0.78]])
# décision sur ces données
y_test = classifier.predict(X_test)
# affichage du résultat
print(y_test)
```

```python
#comment afficher ces points dans le graphique ?
plt.scatter (X[:, 0], X[:, 1], c=y, cmap=my_cmap, edgecolors='gray')
plt.scatter (X_test[:, 0], X_test[:, 1], c=y_test, cmap=my_cmap, edgecolors='yellow', s=150, marker='D')
plt.xlabel (("x_1"))
plt.ylabel (("x_2"))
plt.legend()
plt.show()
```
